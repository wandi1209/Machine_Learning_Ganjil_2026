## 1. Tindakan yang Melanggar Etika dan Hukum dalam Penggunaan AI

Salah satu contoh penyalahgunaan AI adalah **menggunakan data pribadi orang lain tanpa izin**, misalnya memasukkan foto, dokumen identitas, atau informasi pribadi seseorang ke dalam layanan AI tanpa persetujuan. Hal tersebut dapat melanggar privasi dan perlindungan data pribadi.

Contoh lainnya adalah menggunakan AI untuk membuat **deepfake atau informasi palsu** yang digunakan untuk menipu, mencemarkan nama baik, atau merugikan orang lain. Penggunaan AI seharusnya tetap memperhatikan keamanan, privasi, transparansi, kemanusiaan, serta hak orang lain. Hal ini juga sejalan dengan **Surat Edaran Menteri Komunikasi dan Informatika Nomor 9 Tahun 2023 tentang Etika Kecerdasan Artifisial**.

### Contoh tindakan yang tidak etis:

- Menggunakan data pribadi tanpa izin.
- Membuat deepfake untuk menipu orang lain.
- Menyebarkan berita atau informasi palsu yang dibuat AI.
- Menggunakan AI untuk melakukan penipuan.
- Menggunakan hasil AI tanpa melakukan pengecekan kebenaran informasi.

## 2. Dampak Energi dan Lingkungan dari AI

Penggunaan AI membutuhkan komputer dan **data center** dengan jumlah komputasi yang besar. Hal ini menyebabkan penggunaan listrik dan kebutuhan pendinginan yang tinggi. Menurut IEA, data center menggunakan sekitar **415 TWh listrik pada 2024**, dan jumlah tersebut diperkirakan meningkat menjadi sekitar **945 TWh pada 2030**. AI menjadi salah satu faktor utama peningkatan kebutuhan listrik tersebut.

Jika listrik yang digunakan berasal dari bahan bakar fosil, peningkatan penggunaan AI dapat menyebabkan **peningkatan emisi karbon** dan memberikan dampak terhadap lingkungan.

Menurut pendapat saya, dampak tersebut dapat dikurangi dengan:

1. Menggunakan model AI yang lebih kecil dan efisien sesuai kebutuhan.
2. Melakukan optimasi model agar membutuhkan lebih sedikit komputasi.
3. Menggunakan sumber energi terbarukan untuk data center.
4. Meningkatkan efisiensi server dan sistem pendingin.
5. Menghindari penggunaan AI secara berlebihan untuk pekerjaan yang sebenarnya sederhana.

Dengan demikian, perkembangan AI tidak harus dihentikan, tetapi perlu diimbangi dengan **efisiensi energi dan penggunaan teknologi yang lebih ramah lingkungan**. IEA juga memperkirakan energi terbarukan akan memenuhi hampir setengah dari pertumbuhan kebutuhan listrik data center hingga 2030.
